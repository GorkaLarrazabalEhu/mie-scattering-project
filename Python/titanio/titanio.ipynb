```python
import numpy as np
import miepython
import matplotlib.pyplot as plt
```

```python
link = "https://refractiveindex.info/tmp/database/data-nk/main/Ti/Johnson.txt"

tiData = np.genfromtxt(link, delimiter = '\t')
N = len(tiData)//2
waveLength = tiData[1:N,0]
realRefractiveInd = tiData[1:N,1]
imRefractiveInd = - tiData[N+1:,1]
refractiveIndex = realRefractiveInd + 1.0j*imRefractiveInd
```

```python
plt.plot(waveLength, realRefractiveInd, "red")
plt.plot(waveLength, - imRefractiveInd, "blue")
plt.show()
```

```python
diameter = 0.1 #um ... 100nm diameter
qext, qsca, qback, g = miepython.ez_mie(refractiveIndex, diameter, waveLength)
plt.plot(waveLength*1000,qsca*(4*np.pi*diameter**2),color='red')
plt.xlabel("Wavelength (nm)")
plt.ylabel("Efficiency $Q_{ext}$")
plt.title("Eficiencia de dispersion del titanio")
plt.show()
```
