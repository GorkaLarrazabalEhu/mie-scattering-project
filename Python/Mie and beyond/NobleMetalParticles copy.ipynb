```python
import numpy as np
import matplotlib.pyplot as plt
import miepython
import time
```

```python
ag = np.genfromtxt("https://refractiveindex.info/tmp/database/data-nk/main/Ag/Johnson.txt", delimiter = '\t')
au = np.genfromtxt("https://refractiveindex.info/tmp/database/data-nk/main/Au/Johnson.txt", delimiter = '\t')
```

```python
N = len(ag)//2
ag_lam = ag[1:N,0]
ag_mre = ag[1:N,1]
ag_mim = ag[N+1:,1]
cmplxRefractionIndexag = ag_mre +1.0j*ag_mim

N = len(au)//2
au_lam = au[1:N,0]
au_mre = au[1:N,1]
au_mim = au[N+1:,1]
cmplxRefractionIndexau = au_mre +1.0j*au_mim
```

```python
d = 90E-3
qext = miepython.ez_mie(cmplxRefractionIndexag, d, ag_lam)[0]
plt.plot(ag_lam,qext,color='red')
qext = miepython.ez_mie(cmplxRefractionIndexau, d, au_lam)[0]
plt.plot(au_lam,qext,color='blue')
plt.xlabel("Wavelength (um)")
plt.ylabel("Efficiency $Q_{ext}$")
plt.title("Eficiencia de extinción Plata")
#plt.xlim(300,800)
#plt.ylim(1,2.5)
plt.show()
```
